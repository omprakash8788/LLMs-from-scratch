Multiline statements in `py` can terminate with a new line or the backslash `("\")`



```python
# Example
x1 =6
x2=6
x3=7
total = x1 + \
        x2 + \
        x3

print(total)
```

```python
# Example
x1 =6; x2=6; x3=7
total = x1 + \
        x2 + \
        x3

print(total)
```
