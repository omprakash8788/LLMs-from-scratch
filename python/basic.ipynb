Multiline statements in `py` can terminate with a new line or the backslash `("\")`



```python
# Example
x1 =6
x2=6
x3=7
total = x1 + \
        x2 + \
        x3

print(total)
```

```python
# Example
x1 =6; x2=6; x3=7
total = x1 + \
        x2 + \
        x3

print(total)
```

A sting literal that beigns with the letter "r" (for row) treats everything as a literal charater and `escapes` 

```python
# Example
a1=r'\n'
a2=r'\r'
a3=r'\t'
print('a1:', a1, 'a2:', a2, 'a3:', a3)
```
